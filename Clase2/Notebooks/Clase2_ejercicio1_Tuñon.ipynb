## Ejercicio N°1 - Importación y visualizacion de archivos .CSV ; .JSON y .XLS con Pandas





```python
import pandas as pd

# Rutas en formato RAW de GitHub
url_clientes = "https://raw.githubusercontent.com/arieltunon74/Aprendizaje_Automatico_2026/main/Clase2/Ejercicio-1/clientes.json"
url_inventario = "https://raw.githubusercontent.com/arieltunon74/Aprendizaje_Automatico_2026/main/Clase2/Ejercicio-1/inventario.xlsx"
url_ventas = "https://raw.githubusercontent.com/arieltunon74/Aprendizaje_Automatico_2026/main/Clase2/Ejercicio-1/ventas.csv"

# Importar archivo JSON (Clientes)
df_clientes = pd.read_json(url_clientes).set_index("id_cliente")
print("--- Clientes ---")
display(df_clientes.head())

# Importar archivo Excel (Inventario)
df_inventario = pd.read_excel(url_inventario)
print("\n--- Inventario ---")
display(df_inventario.head())

# Importar archivo CSV (Ventas)
df_ventas = pd.read_csv(url_ventas, index_col="id_venta")
print("\n--- Ventas ---")
display(df_ventas.head())
```
