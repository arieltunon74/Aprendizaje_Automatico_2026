```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2, 100) # Corrected 'O' to '0' and 'lee' to '100'

plt.plot(x, x, 'o', label='Linear') # Corrected invalid characters and added a marker
plt.plot(x, x**2, '-', label='Quadratic') # Corrected invalid characters and added a line style
plt.plot(x, x**3, ':', label='Cubic') # Corrected invalid characters and added a line style

plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.title('Título del gráfico') # Corrected 'de gråfico"' to a valid title command
plt.legend() # Added legend to display labels
plt.show() # Corrected 'Show()' to 'show()'
```
